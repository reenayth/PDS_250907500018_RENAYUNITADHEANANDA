# Regresi

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

data = {
    'Partisipan': [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    'PMakan': [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    'PTranspor': [20, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    'BB': [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]}

df = pd.DataFrame(data)

X = df[['PMakan', 'PTranspor']]
y = df['BB']

X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()

print(results.summary())
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
dataa2 = pd.read_csv(url)

print(dataa2.shape)
```

```python
X = dataa2[['DBH_PAJAK', 'DAU_Grant', 'DAK_Nonfisik']]
y = dataa2['PAD']

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

model = LinearRegression()
model.fit(X, y)
```

```python
b0 = model.intercept_
b1, b2, b3 = model.coef_

y_hat = model.predict(X)
r2 = r2_score(y, y_hat)

n, k = len(dataa2), 3
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - k - 1)

print(f"Y = {b0:.4f} + {b1:.4f} X1 + {b2:.4f} X2 + {b3:.4f} X3")
print(f"R2 = {r2:.4f}")
print(f"Adjusted R2 = {adj_r2:.4f}")
```

```python
prediksi = pd.DataFrame([[77.91, 96.98, 72.02]], columns=X.columns)
print("Prediksi Y:", model.predict(prediksi)[0].round(2))
```

```python
import plotly.express as px

fig = px.scatter_3d(dataa2, x='DBH_PAJAK', y='DAU_Grant', z='PAD',
                    color='DAK_Nonfisik',
                    title='Regresi Linier 3D: X1, X2, Y (warna = X3)')
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
dataa2 = pd.read_csv(url)

X = dataa2[['DAU_Grant']].values
Y = dataa2['PAD'].values

model = LinearRegression().fit(X, Y)
print(f"ŷ = {model.intercept_:.3f} + {model.coef_[0]:.3f}X | R² = {model.score(X, Y):.3f}")

plt.scatter(X, Y, s=80, label="Data Aktual")
plt.plot(X, model.predict(X), linewidth=2, label="Garis Regresi")
plt.xlabel("DAU_Grant")
plt.ylabel("PAD")
plt.title("Regresi Linear: DAU_Grant terhadap PAD")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
dataa2 = pd.read_csv(url)
dataa2

# Menampilkan data Kota Lubuk Linggau
data_lubuk = dataa2[
    dataa2["Kota"].astype(str).str.contains(
        "Lubuk Linggau", case=False, na=False
    )
]

# Menampilkan DAU Grant, DAU Earmark, DAK Fisik, dan Pengangguran
kolom = [
    "Kota", "Tahun", "DAU_Grant",
    "DAU_Earmark", "DAK_Fisik", "PENGANGGURAN"
]

display(data_lubuk[kolom].sort_values("Tahun"))
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

data = {
    "Tahun": [2014, 2015, 2016, 2017, 2018, 2019,
              2020, 2021, 2022, 2023, 2024],
    "DAU_Grant": [92.89, 101.45, 86.82, 94.25, 74.41,
                  99.77, 101.19, 112.42, 93.35, 93.24, 99.47],
    "DAU_Earmark": [55.17, 58.40, 71.89, 49.51, 63.10,
                    45.99, 82.88, 49.02, 47.86, 41.89, 62.43],
    "DAK_Fisik": [52.46, 63.04, 42.50, 68.62, 48.81,
                  50.74, 60.80, 62.28, 80.10, 63.70, 52.07],
    "PENGANGGURAN": [7, 8, 8, 7, 6, 9, 6, 7, 5, 6, 6]
}

data3 = pd.DataFrame(data)
display(data3)

#Variabel independen
x = data3[["DAU_Grant", "DAU_Earmark", "DAK_Fisik"]]

#Variabel dependen
Y = data3["PENGANGGURAN"]

#Intercept
X_model = sm.add_constant(x)

#Estimasi model ols
model = sm.OLS(Y, X_model)
results = model.fit()

#Menampilkan ringkasan hasil regresi
print("\n")
print("=" * 70)
print("HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())

#Mengambil koefisien regresi
intercept = results.params["const"]
coef_grant = results.params["DAU_Grant"]
coef_earmark = results.params["DAU_Earmark"]
coef_fisik = results.params["DAK_Fisik"]

r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

#Membuat persamaan regresi
persamaan = (
f"Ŷ = {intercept:.3f}"
f" + ({coef_grant:.3f}) x1"
f" + ({coef_earmark:.3f}) x2"
f" + ({coef_fisik:.3f}) x3")

print("\n")
print("=" * 70)
print("PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

#Menghitung nilai prediksi
data3["Prediksi"] = results.predict(X_model)
print("\n")
print("=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(
data3[["Tahun", "DAU_Grant", "DAU_Earmark", "DAK_Fisik",
]].round(2))

#Membuat grid untuk bidang regresi
x1 = np.linspace(
    data3["DAU_Grant"].min(),
    data3["DAU_Grant"].max(),
    40)

x2 = np.linspace(
    data3["DAU_Earmark"].min(),
    data3["DAU_Earmark"].max(),
    40)

x3 = np.linspace(
    data3["DAK_Fisik"].min(),
    data3["DAK_Fisik"].max(),
    40)
x1, x2, x3 = np.meshgrid(x1, x2, x3)

#Menghitung nilai Y pada bidang regresi
Y = (intercept + coef_grant * x1 + coef_earmark * x2 + coef_fisik * x3)

#Membuat figure 3D
fig = go.Figure()

#Menampilkan data aktual
fig.add_trace(
    go.Scatter3d(
        x=data3["DAU_Grant"],
        y=data3["DAU_Earmark"],
        z=data3["PENGANGGURAN"],
        mode="markers",
        marker=dict(
            size=7,
            color="red",
            opacity=0.9
        ),

        name="Data Aktual",
        text=[
            f"Partisipan {p}"
            for p in data3["Tahun"]
        ],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "DAU_Grant: %{x}<br>"
            "DAU_Earmark: %{y}<br>"
            "PENGANGGURAN: %{z}"
            "<extra></extra>"
        )
    )
)



# 14. MENAMPILKAN BIDANG REGRESI
fig.add_trace(
    go.Surface(
        x=x1, y=x2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Regression Plane",
        hovertemplate=(
            "DAU_Grant: %{x:.2f}<br>"
            "DAU_Earmark: %{y:.2f}<br>"
            "Prediksi Y: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

#Mengatur tampilan grafik
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D"
            "<br>"
            f"<sup>{persamaan} | "
            f"R2 = {r_squared:.3f}</sup>" ), "x": 0.5 },
    scene={
        "xaxis": {"title": "Dana Alokasi Umum Grant (x1)" },
        "yaxis": {"title": "Dana Alokasi Umum Earmark (x2)" },
        "zaxis": { "title": "Jumlah Pengangguran (Y)" } },
    width=1000,
    height=750,
    margin={
        "l": 0, "r": 0, "b": 0, "t": 100 },
    legend={ "x": 0.02, "y": 0.98 }
)

fig.show()
```

```python
import numpy as np
from scipy import stats
data = np.array([167, 184, 192, 198, 200, 202, 210, 211, 212, 215, 216, 217,])

# Perform normality test
result = stats.normaltest(data)

# Extract test statistics and p-value
test_statistic = result.statistic
p_value = result.pvalue

# Print the results
print("Test Statistic:", test_statistic)
print("P-Value:", p_value)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

data = np.array([167, 184, 192, 198, 200, 202, 210, 211, 212, 215, 216, 217,])

#Plot histogram
plt.hist(data, bins='auto', density=True, alpha=0.7, color='skyblue', edgecolor='black')

# Plot kurva distribusi normal
mu, std = np.mean(data), np.std(data)
x = np.linspace(np.min(data), np.max(data), 100)
y = stats.norm.pdf(x, mu, std)
plt.plot(x, y, color='red')

# Set plot title and labels
plt.title('Histogram of Data with Normal Distribution')
plt.xlabel('Data')
plt.ylabel('Frequency')

# Show the plot
plt.show()

# QQ plot
plt.figure(figsize=(6, 4))
stats.probplot(data, dist="norm", plot=plt)

# Set plot title and labels
plt.title('QQ Plot')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Sample Quantiles')

# Show the plot
plt.show()
```
