```python
import seaborn as sns
import matplotlib.pyplot as plt

# data
bulan = ['jan', 'feb', 'mar', 'april', 'mei', 'juni', 'juli']
penjualan = [ 150, 200, 180, 250, 125, 220, 130]

plt.figure(figsize=(8, 4))

sns.barplot(x=bulan, y=penjualan, palette='pastel')

plt.plot(
    range(len(bulan)),
    penjualan,
    color='red',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='garis tren'
)

plt.title('penjualan per bulan dengan garis tren', fontsize=12, fontweight='bold')
plt.xlabel('bulan')
plt.ylabel('jumlah penjualan')
plt.legend()

plt.tight_layout()
plt.show()
```

# Dataset 1

```python
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

dataa1 = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv')
dataa1
```

```python
dataa1['Total'] = dataa1[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

plt.figure(figsize=(10, 5))
plt.bar(dataa1['nama'], dataa1['Total'], color='royalblue', edgecolor='black')

plt.title('Total Penjualan per Orang (Januari - April)', fontsize=14)
plt.xlabel('Nama', fontsize=12)
plt.ylabel('Total Penjualan', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

```python
# Pie Chart
plt.figure(figsize=(5, 5))

total_per_bulan = dataa1[['jan', 'feb', 'mar', 'apr']].sum()

plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
        startangle=140, colors=sns.color_palette('muted'))

plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# stack plot
plt.figure(figsize=(8,5))

dataa1.set_index('nama')[['jan', 'feb', 'mar', 'apr']].plot(
    kind='bar',
    stacked=True,
    colormap='Pastel1',
    ax=plt.gca()
)

plt.title('Komposisi Penjualan Bulanan per Orang (Stacked)', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Jumlah Penjualan')
plt.xticks(rotation=45)
plt.legend(title='Bulan')
plt.tight_layout()
plt.show()
```

```python
# Histogran
plt.figure(figsize=(6, 4))

sns.histplot(dataa1['Total'], bins=5, kde=True, color='teal')

plt.title('Distribusi Frekuensi Total Penjualan', fontsize=12)
plt.xlabel('Range Total Penjualan')
plt.ylabel('Jumlah Karyawan (Frekuensi)')
plt.tight_layout()
plt.show()
```

```python
# scatter
plt.figure(figsize=(6,4))

sns.scatterplot(data=dataa1, x='jan', y='feb', s=100, color='white', hue='nama')

for i in range(dataa1.shape[0]):
  plt.text(dataa1['jan'].iloc[i] + 5, dataa1['feb'].iloc[i], dataa1['nama'].iloc[i])

plt.title('Korelasi Penjualan: Januari vs Februari', fontsize=12)
plt.xlabel('Penjualan Januari')
plt.ylabel('Penjualan Februari')
plt.tight_layout()
plt.show()
```

```python
# Time Series Plot
plt.figure(figsize=(7, 4))

rata_bulanan = dataa1[['jan', 'feb', 'mar', 'apr']].mean()

sns.lineplot(x=rata_bulanan.index, y=rata_bulanan.values, marker='o')

plt.title('Rata-rata penjualan tim (Time Series)', fontsize=12)
plt.xlabel('Bulan')
plt.ylabel('Rata-rata Penjualan')
plt.tight_layout()
plt.show()
```
